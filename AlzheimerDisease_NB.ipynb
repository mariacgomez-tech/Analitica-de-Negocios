# **Caso de estudio: Detección de riesgo de Alzheimer**
En este documento se desarrolla y analiza un modelo Naive Bayes para la identificación
temprana de pacientes con diagnóstico de Alzheimer, basándose en variables clínicas,
demográficas y de estilo de vida del paciente.

De acuerdo con lo anterior, las variables utilizadas son todas las variables numericas del dataset para mayor precisión del modelo:

**Age**: Edad del paciente en años al momento del estudio.

**Gender**: Género del paciente (0 = Femenino, 1 = Masculino).

**Ethnicity**: Etnia del paciente (0 = Caucásico, 1 = Afroamericano, 2 = Asiático, 3 = Otro).

**EducationLevel**: Nivel educativo del paciente (0 = Ninguno, 1 = Secundaria, 2 = Universitario, 3 = Posgrado).

**BMI**: Índice de masa corporal del paciente.

**Smoking**: Indica si el paciente fuma (0 = No, 1 = Sí).

**AlcoholConsumption**: Consumo semanal de alcohol en unidades.

**PhysicalActivity**: Actividad física semanal en horas.

**DietQuality**: Puntuación de la calidad de la dieta del paciente (0-10).

**SleepQuality**: Puntuación de la calidad del sueño del paciente (4-10).

**FamilyHistoryAlzheimers**: Indica si hay antecedentes familiares de Alzheimer (0 = No, 1 = Sí).

**CardiovascularDisease**: Indica si el paciente tiene enfermedad cardiovascular (0 = No, 1 = Sí).

**Diabetes**: Indica si el paciente tiene diabetes (0 = No, 1 = Sí).

**Depression**: Indica si el paciente tiene depresión (0 = No, 1 = Sí).

**HeadInjury**: Indica si el paciente ha sufrido una lesión en la cabeza (0 = No, 1 = Sí).

**Hypertension**: Indica si el paciente tiene hipertensión (0 = No, 1 = Sí).

**SystolicBP**: Presión arterial sistólica del paciente (mmHg).

**DiastolicBP**: Presión arterial diastólica del paciente (mmHg).

**CholesterolTotal**: Nivel de colesterol total del paciente (mg/dL).

**CholesterolLDL**: Nivel de colesterol LDL del paciente (mg/dL).

**CholesterolHDL**: Nivel de colesterol HDL del paciente (mg/dL).

**CholesterolTriglycerides**: Nivel de triglicéridos del paciente (mg/dL).

**MMSE**: Puntuación del Mini-Mental State Examination. Evalúa la función cognitiva del paciente (0-30).

**FunctionalAssessment**: Evaluación de la capacidad funcional del paciente (0-10).

**MemoryComplaints**: Indica si el paciente reporta quejas de memoria (0 = No, 1 = Sí).

**BehavioralProblems**: Indica si el paciente presenta problemas de comportamiento (0 = No, 1 = Sí).

**ADL**: Puntuación de actividades de la vida diaria del paciente (0-10).

**Confusion**: Indica si el paciente presenta episodios de confusión (0 = No, 1 = Sí).

**Disorientation**: Indica si el paciente presenta desorientación (0 = No, 1 = Sí).

**PersonalityChanges**: Indica si el paciente presenta cambios de personalidad (0 = No, 1 = Sí).

**DifficultyCompletingTasks**: Indica si el paciente tiene dificultad para completar tareas (0 = No, 1 = Sí).

**Forgetfulness**: Indica si el paciente presenta olvidos frecuentes (0 = No, 1 = Sí).

0. Se procede con la carga de las librerías de trabajo

```python
import numpy as np
import pandas as pd

from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix
```

1. Se cargan los datos de trabajo de la base de datos Altzheimer disease

```python
nxl='/content/Alzheimer_disease_dataset.xlsx'
XDB=pd.read_excel(nxl,sheet_name=0)
XDB.head(100)

XD = XDB[['Age', 'Gender', 'Ethnicity', 'EducationLevel', 'BMI', 'Smoking',
           'AlcoholConsumption', 'PhysicalActivity', 'DietQuality', 'SleepQuality',
           'FamilyHistoryAlzheimers', 'CardiovascularDisease', 'Diabetes', 'Depression',
           'HeadInjury', 'Hypertension', 'SystolicBP', 'DiastolicBP', 'CholesterolTotal',
           'CholesterolLDL', 'CholesterolHDL', 'CholesterolTriglycerides', 'MMSE',
           'FunctionalAssessment', 'MemoryComplaints', 'BehavioralProblems', 'ADL',
           'Confusion', 'Disorientation', 'PersonalityChanges',
           'DifficultyCompletingTasks', 'Forgetfulness']]
XD.head(10)

yd = XDB['Diagnosis']
yd.head()
```

2. Implementamos modelo Naive Bayes

```python
np.set_printoptions(precision=3)
mnb = GaussianNB()
mnb.fit(XD, yd)

u     = mnb.theta_
sigma = mnb.var_; sigma = np.sqrt(sigma)

print("'Age','Gender','Ethnicity','EducationLevel','BMI','Smoking','AlcoholConsumption',")
print("'PhysicalActivity','DietQuality','SleepQuality','FamilyHistoryAlzheimers',")
print("'CardiovascularDisease','Diabetes','Depression','HeadInjury','Hypertension',")
print("'SystolicBP','DiastolicBP','CholesterolTotal','CholesterolLDL','CholesterolHDL',")
print("'CholesterolTriglycerides','MMSE','FunctionalAssessment','MemoryComplaints',")
print("'BehavioralProblems','ADL','Confusion','Disorientation','PersonalityChanges',")
print("'DifficultyCompletingTasks','Forgetfulness'")
print()
print("Medias por categoría (fila 0 = Sin Alzheimer, fila 1 = Con Alzheimer):")
print(u)
print("Las desviaciones son:")
print(sigma)
print()
print("Los límites superiores de las variables son: ", u + sigma)
print()
print("Los límites inferiores de las variables son: ", u - sigma)
```

**Análisis de Resultados**

De los resultados arrojados por el modelo, podemos observar que las variables con
mayores diferencias entre la categoría sin Alzheimer y con Alzheimer son
FunctionalAssessment, ADL, MMSE, MemoryComplaints y BehavioralProblems, lo que
confirma que estas son las variables más determinantes para la clasificación.

La variable FunctionalAssessment presenta una media de 5.86 en pacientes sin Alzheimer
frente a 3.65 en pacientes con Alzheimer, lo que indica que una menor capacidad
funcional está fuertemente asociada al diagnóstico positivo. De manera similar, la
variable ADL muestra una media de 5.71 en el grupo sin Alzheimer frente a 3.66 en el
grupo con Alzheimer, evidenciando que los pacientes diagnosticados presentan una
mayor dificultad para realizar actividades de la vida diaria.

La variable MMSE, que evalúa la función cognitiva del paciente, registra una media de
16.27 en el grupo sin Alzheimer y de 11.99 en el grupo con Alzheimer, reflejando un
deterioro cognitivo significativamente mayor en los pacientes diagnosticados.

Las variables MemoryComplaints y BehavioralProblems también presentan diferencias
notables entre categorías. MemoryComplaints registra una media de 0.116 en el grupo
sin Alzheimer frente a 0.376 en el grupo con Alzheimer, mientras que BehavioralProblems
muestra 0.096 versus 0.267 respectivamente, lo que indica que los pacientes
diagnosticados reportan con mayor frecuencia quejas de memoria y problemas de
comportamiento.

Las demás variables como Age, Gender, BMI, SystolicBP, DiastolicBP, CholesterolTotal,
CholesterolLDL, CholesterolHDL, Smoking, AlcoholConsumption, PhysicalActivity y
DietQuality no presentan diferencias significativas entre las categorías, lo que indica
que estas características por sí solas no son determinantes para la clasificación del
diagnóstico de Alzheimer.

De acuerdo con los resultados obtenidos, se puede observar de manera general que el
modelo alcanzó una exactitud del 81.3%, lo que indica un buen desempeño en la
clasificación de los pacientes entre las categorías con y sin diagnóstico de Alzheimer.

Se destaca la especificidad del 86.8%, lo que indica que el modelo reconoce
correctamente a la gran mayoría de los pacientes que realmente no presentan diagnóstico
de Alzheimer. De igual manera, la predicción negativa del 84.7% confirma que la mayoría
de los pacientes clasificados como sin Alzheimer realmente no lo tienen.

La sensibilidad del 71.4% indica que el modelo logra identificar correctamente a un
porcentaje significativo de los pacientes que sí presentan diagnóstico de Alzheimer,
detectando 543 de los 760 casos positivos de la base de datos. La precisión del 74.7%
evidencia que cuando el modelo clasifica a un paciente como con Alzheimer, en la gran
mayoría de los casos el diagnóstico es correcto.

En comparación con el modelo aplicado al dataset de Stroke, este modelo presenta un
desempeño notablemente superior en todas las métricas, especialmente en sensibilidad,

(64.6% sin Alzheimer vs 35.4% con Alzheimer).
