Se quiere implementar un modelo de árbol de decisión para pronosticar si los pacientes
que hacen parte de esta base de datos van a ser diagnosticados con Alzheimer.

0. Se procede con la carga de librerías de trabajo

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import confusion_matrix
```

1. Se procede con la carga de datos de trabajo.

```python
nxl= '/content/Alzheimer_disease_dataset.xlsx'
XDB= pd.read_excel(nxl, sheet_name= 0)

#Para el modelo de árbol de decisión se seleccionaron las 5 variables con mayor
#correlación con el diagnóstico, identificadas previamente en el modelo Naive Bayes

XD1 = XDB[['FunctionalAssessment', 'ADL', 'MemoryComplaints', 'MMSE', 'BehavioralProblems']]
yd1 = XDB['Diagnosis']

display(XD1)
display(yd1)
```

2. Se procede con la implementación del modelo de árbol

```python
mar=DecisionTreeClassifier(criterion='gini', max_depth=4) #Entrenamos el árbol con base histórica de datos
mar.fit(XD1,yd1)

ydp1=mar.predict(XD1) #Predicción del modelo
#Se construye la matriz de confusión
cm=confusion_matrix(yd1,ydp1)
display(cm)
VN=cm[0,0]; FP=cm[0,1]; FN=cm[1,0]; VP=cm[1,1]

#Métricas de desempeño
Ex=(VP+VN)/len(XD1) #1. Exactitud: Comportamiento general
print('La Exactitud es: ',Ex)
Sen=VP/(VP+FN) #2. Sensibilidad: Como se comporta pronosticando PreAprobados
print('La Sensibilidad es: ',Sen)
Spe=VN/(VN+FP) #3. Especificidad: Como se comporta frente a los negativos
print('La Especificidad es: ',Spe)
Pre=VP/(VP+FP) #4. Precisión: Como se comporta pronosticando PreAprobados
print('La Precisión es: ',Pre)
PreNeg=VN/(VN+FN) #5. Precisión Negativa: Como se comporta pronosticando Negativos
print('La Precisión Negativa es: ',PreNeg)
```

**Análisis de resultados**

De la base de datos histórica se observa un total de 2149 registros, de los cuales
1389 corresponden a pacientes sin diagnóstico de Alzheimer y 760 corresponden a
pacientes con diagnóstico de Alzheimer.

De los 1389 casos sin Alzheimer, el modelo clasificó correctamente 1335 pacientes,
mientras que de los 760 casos con Alzheimer, el modelo identificó correctamente 659
registros. El modelo logra identificar en un 96.1% a los pacientes sin Alzheimer
(1335/1389), mientras que logra identificar en un 86.7% a los pacientes con Alzheimer
(659/760).

Con respecto a las métricas, el modelo alcanzó una exactitud del 92.8%, lo que indica
un excelente comportamiento general del modelo frente a la clasificación en ambas
categorías. Se destacan especialmente la especificidad (96.1%) y la precisión negativa
(93.0%), las cuales se ubican ampliamente por encima del umbral del 75% definido como
referencia para este tipo de modelos de clasificación. La sensibilidad (86.7%) también
presenta un desempeño sólido, mostrando una adecuada capacidad para detectar pacientes
con diagnóstico de Alzheimer. La precisión (92.4%) evidencia que cuando el modelo
clasifica a un paciente como con Alzheimer, en la gran mayoría de los casos el
diagnóstico es correcto.

Es importante destacar que este modelo, construido únicamente con las 5 variables más
determinantes (FunctionalAssessment, ADL, MemoryComplaints, MMSE y BehavioralProblems),
supera en todas las métricas al modelo Naive Bayes que utilizó las 32 variables
disponibles, lo que confirma que estas variables capturan de manera efectiva los
patrones de clasificación del diagnóstico de Alzheimer.


3. Despliegue del árbol de decisión

```python
from sklearn.tree import export_graphviz #Exporta los datos a un gráfico
from pydotplus import graph_from_dot_data #Es un graficador

vs=['FunctionalAssessment', 'ADL', 'MemoryComplaints', 'MMSE', 'BehavioralProblems'] #Títulos del árbol
dot_data=export_graphviz(mar, feature_names=vs) #Exportar de números a gráfico en pdf
graph=graph_from_dot_data(dot_data) #Graficamo el árbol
graph.write_png('arbol_alzheimer.png') #Guardamos el árbol en pdf
```

**Análisis de resultados**
De acuerdo con el árbol de decisión, la variable raíz es FunctionalAssessment con un
umbral de 5.001, dividiendo los 2149 pacientes en dos grandes grupos: 1054 pacientes
con baja capacidad funcional (rama izquierda) y 1095 con alta capacidad funcional
(rama derecha), lo que confirma que esta variable es el factor más determinante para
la clasificación del diagnóstico de Alzheimer.

En la rama derecha del árbol se identifica un nodo puro con value = [24, 0], donde el 100% de los individuos son
clasificados como sin Alzheimer, evidenciando un perfil de bajo riesgo claramente
diferenciado. Así mismo, se destaca el nodo casi puro con value = [710, 27] con un gini de 0.071, donde la gran
mayoría de los individuos no presentan diagnóstico de Alzheimer.

En la rama izquierda se destaca el nodo casi puro con value = [158, 3] y gini = 0.037, donde casi la totalidad
de los individuos no presentan Alzheimer. Por otro lado, el nodo con value = [2, 5]
y gini = 0.408, concentra la mayor proporción de casos positivos, evidenciando
que la combinación de baja capacidad funcional, dificultad en actividades diarias
y quejas de memoria constituye el perfil de mayor riesgo para el diagnóstico
de Alzheimer.

4. Pronósticar con 'Datos Pronóstico'

```python
#Paciente 1 → [4.0, 4.0, 1, 18, 1]
#Paciente 2 → [8.0, 9.0, 0, 27, 0]
#Paciente 3 → [5.5, 5.0, 1, 22, 1]

XD2 = pd.DataFrame({
    'FunctionalAssessment': [4.0, 8.0, 5.5],
    'ADL':                  [4.0, 9.0, 5.0],
    'MemoryComplaints':     [1,   0,   1  ],
    'MMSE':                 [18,  27,  22 ],
    'BehavioralProblems':   [1,   0,   1  ]
})

ydp2 = mar.predict(XD2)
display(ydp2)

for i, pred in enumerate(ydp2):
    if pred == 1:
        print(f"Paciente {i+1}: El paciente tiene diagnóstico positivo de Alzheimer")
    else:
        print(f"Paciente {i+1}: El paciente no presenta diagnóstico de Alzheimer")
```

**Análisis Pronóstico**

De los 3 pacientes evaluados, el modelo pronostica que 1 paciente (1/3 = 33.3%)
no presenta diagnóstico de Alzheimer, mientras que 2 pacientes (2/3 = 66.7%)
tienen diagnóstico positivo.
